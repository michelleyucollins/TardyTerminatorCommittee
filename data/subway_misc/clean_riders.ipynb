```python
for i in [1, 2, 4]:
    # Read the CSV file into a DataFrame
    df = pd.read_csv(f"line_{i}_ridership.csv")

    # Capitalize all the Station names
    df['Station'] = df['Station'].str.upper()

    # Remove all '.' characters from the Station names
    df['Station'] = df['Station'].str.replace('.', '', regex=False)

    # Add 'STATION' to the end of each Station name
    df['Station'] = df['Station'] + ' STATION'

    # update the CSV file with the modified DataFrame
    df.to_csv(f"line_{i}_ridership.csv", index=False)
```
